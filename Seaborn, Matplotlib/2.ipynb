# Stripplot :
A stripplot is essentially a basic scatter plot where one of the axes represents categories. It draws a single dot for every data point.The Overlap Problem: Because points in the same category with similar values are plotted on the exact same axis line, they can easily overlap, making it hard to tell how many points are actually there.The Fix: To solve this, stripplots often use jitter (adding a tiny bit of random horizontal noise) to spread the points out slightly so you can see the density better.

# Swarmplot
A swarmplot (also known as a beeswarm plot) is similar to a stripplot, but it automatically uses an algorithm to arrange the points so that they never overlap.

Better Density Visualization: Because the points are neatly stacked next to each other horizontally, the resulting shape outlines the actual distribution of the data (similar to a violin plot).

The Limitation: Swarmplots do not scale well to very large datasets. If you have thousands of data points, the algorithm struggles to arrange them without overlapping or stretching the plot far beyond the category bounds.

```python
import seaborn as sns
import matplotlib.pyplot as plt
df=sns.load_dataset("tips")
df
```

```python
sns.stripplot(x='day',y='total_bill',data=df)
```

```python
sns.swarmplot(x='day',y='total_bill',data=df)
```

```python
sns.violinplot(x='tip',y='day',data=df)
sns.swarmplot(x='tip',y='day',data=df)
```
