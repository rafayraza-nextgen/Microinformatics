# Categorical Data Plots
## Now let's discuss using seaborn to plot categorical data! There are a few main plot types for this:

- boxplot

- violinplot

- stripplot

- swarmplot

- barplot

- countplot

```python
df=sns.load_dataset('tips')
df
```

```python
sns.countplot(df['sex'])
```

```python
sns.countplot(x=df['sex'])
```

```python
sns.countplot(x=df['sex'],hue=df['smoker'])
```

```python
sns.barplot(x=df['tip'])
```

```python
sns.barplot(x=df['sex'],y=df['total_bill'])
```

## You can change the estimator object to your own function, that converts a vector to a scalar

Mean (default): np.mean

Median: np.median

Sum: np.sum

Count: len (to count the number of observations)

Standard Deviation: np.std

Minimum: np.min

Maximum: np.max

```python
sns.barplot(x=df['sex'],y=df['total_bill'],estimator=np.sum)
```

```python
sns.barplot(x=df['sex'],y=df['tip'],estimator=np.sum)
```

```python
sns.boxplot(x='tip',y='day',data=df)
```

```python
sns.boxplot(x='tip',y='day',data=df,palette='rainbow')
```

# The Box Plot
A box plot (or box-and-whisker plot) provides a clean, standardized way to display a five-number statistical summary of your dataset. It is highly effective for identifying outliers and understanding the basic spread of your data.

# The Violin Plot
A violin plot plays a similar role but packs in more visual information. It essentially combines a miniature box plot with a rotated Kernel Density Estimation (KDE) plot on each side.

```python
sns.violinplot(x='day',y='total_bill',data=df)
```
